```python
#include <iostream>
#include <vector>
#include <iomanip>
#include <numeric>

using namespace std;

int main()
{
    vector<double> diem = {7.5, 4.0, 8.5, 9.0, 5.5};

    double tong = accumulate(diem.begin(), diem.end(), 0.0);

    double diemTB = tong / diem.size();

    vector<double> dat;

    for (double d : diem)
    {
        if (d >= 5)
        {
            dat.push_back(d);
        }
    }


    cout << fixed << setprecision(2);

    cout << "Diem TB: " << diemTB << endl;

    cout << "dat: [";

    for (int i = 0; i < dat.size(); i++)
    {
        cout << dat[i];

        if (i < dat.size() - 1)
        {
            cout << ", ";
        }
    }

    cout << "]" << endl;

    return 0;
}
```
